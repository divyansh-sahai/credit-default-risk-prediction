# Phase 2: Data Cleaning

## Missing Value Analysis

```python
high_missing_cols = missing_table[
    missing_table['Missing_Percentage'] > 50
]

print("Columns with >50% missing values:")
print(len(high_missing_cols))
```

```python
high_missing_cols
```

```python
cols_to_drop = high_missing_cols.index.tolist()

print("Columns selected for removal:")
print(len(cols_to_drop))
```

```python
df_clean = df.drop(columns=cols_to_drop)

print("Original Shape :", df.shape)
print("New Shape      :", df_clean.shape)
```

```python
numerical_cols = df_clean.select_dtypes(
    include=['int64', 'float64']
).columns

len(numerical_cols)
```

```python
print("Categorical Columns")
print(list(categorical_cols))
```

## Missing Value Treatment

```python
for col in numerical_cols:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].median()
    )

print("Numerical missing values handled.")
```

```python
for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].mode()[0]
    )

print("Categorical missing values handled.")
```

# Phase 5: Model Development

## Data Preparation for Modeling

```python
model_df = df_clean.copy()

model_df.drop(
    columns=['SK_ID_CURR'],
    inplace=True
)
```

```python
model_df.select_dtypes(
    include='object'
).columns.tolist()
```

```python
model_df = pd.get_dummies(
    model_df,
    drop_first=True
)
```

```python
model_df.shape
```

```python
model_df.select_dtypes(
    include='object'
).shape
```

```python
X = model_df.drop(
    columns=['TARGET']
)

y = model_df['TARGET']

print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
y_train.mean() * 100
```

```python
y_test.mean() * 100
```

## Model 1: Logistic Regression

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model.fit(
    X_train_scaled,
    y_train
)

print("Logistic Regression Trained")
```

```python
y_pred_log = log_model.predict(X_test_scaled)

y_prob_log = log_model.predict_proba(
    X_test_scaled
)[:,1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Accuracy :", accuracy_score(y_test, y_pred_log))
print("Precision:", precision_score(y_test, y_pred_log))
print("Recall   :", recall_score(y_test, y_pred_log))
print("F1 Score :", f1_score(y_test, y_pred_log))
print("ROC AUC  :", roc_auc_score(y_test, y_prob_log))
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(
    y_test,
    y_pred_log
)
```

```python
from sklearn.linear_model import LogisticRegression

log_model_balanced = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

log_model_balanced.fit(
    X_train_scaled,
    y_train
)
```

```python
y_pred_bal = log_model_balanced.predict(
    X_test_scaled
)

y_prob_bal = log_model_balanced.predict_proba(
    X_test_scaled
)[:,1]
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred_bal))
print("Precision:", precision_score(y_test, y_pred_bal))
print("Recall   :", recall_score(y_test, y_pred_bal))
print("F1 Score :", f1_score(y_test, y_pred_bal))
print("ROC AUC  :", roc_auc_score(y_test, y_prob_bal))
```

```python
confusion_matrix(
    y_test,
    y_pred_bal
)
```

```python
from scipy.stats import ks_2samp

ks_stat = ks_2samp(
    y_prob_bal[y_test == 0],
    y_prob_bal[y_test == 1]
)

ks_stat
```

## Model 3: XGBoost

```python
import xgboost
print(xgboost.__version__)
```

## XGBoost Model Training

```python
negative_class = (y_train == 0).sum()
positive_class = (y_train == 1).sum()

scale_pos_weight = negative_class / positive_class

print("Scale Pos Weight:", scale_pos_weight)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    eval_metric='logloss'
)
```

```python
xgb_model.fit(
    X_train,
    y_train
)

print("XGBoost Trained Successfully")
```

```python
y_pred_xgb = xgb_model.predict(X_test)

y_prob_xgb = xgb_model.predict_proba(
    X_test
)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred_xgb))
print("Precision:", precision_score(y_test, y_pred_xgb))
print("Recall   :", recall_score(y_test, y_pred_xgb))
print("F1 Score :", f1_score(y_test, y_pred_xgb))
print("ROC AUC  :", roc_auc_score(y_test, y_prob_xgb))
```

```python
confusion_matrix(
    y_test,
    y_pred_xgb
)
```

# Phase 7: Risk Segmentation

```python
risk_df = pd.DataFrame({
    'Actual_Default': y_test,
    'Default_Probability': y_prob_xgb
})

risk_df.head()
```

```python
risk_df['Risk_Bucket'] = pd.qcut(
    risk_df['Default_Probability'],
    q=4,
    labels=[
        'Low Risk',
        'Medium Risk',
        'High Risk',
        'Very High Risk'
    ]
)
```

```python
risk_bucket_summary = (
    risk_df['Risk_Bucket']
    .value_counts()
    .reset_index()
)

risk_bucket_summary.columns = [
    'Risk_Bucket',
    'Customer_Count'
]

risk_bucket_summary
```

```python
bucket_default_rate = (
    risk_df.groupby('Risk_Bucket')
    ['Actual_Default']
    .mean()
    * 100
)

bucket_default_rate = (
    bucket_default_rate
    .reset_index()
)

bucket_default_rate.columns = [
    'Risk_Bucket',
    'Default_Rate'
]

bucket_default_rate
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=bucket_default_rate,
    x='Risk_Bucket',
    y='Default_Rate'
)

plt.title('Default Rate by Risk Bucket')
plt.xlabel('Risk Segment')
plt.ylabel('Default Rate (%)')

plt.show()
```

# Phase 8: Credit Decision Framework

```python
underwriting_framework = pd.DataFrame({
    'Risk_Bucket': [
        'Low Risk',
        'Medium Risk',
        'High Risk',
        'Very High Risk'
    ],
    'Decision': [
        'Approve',
        'Approve with Conditions',
        'Manual Review',
        'Reject'
    ]
})

underwriting_framework
```

```python
risk_decision = pd.merge(
    bucket_default_rate,
    underwriting_framework,
    on='Risk_Bucket'
)

risk_decision
```

# Key Business Findings

### Key Drivers of Default Risk

- External credit bureau scores were the strongest predictors of default.
- Employment stability was an important determinant of repayment behavior.
- Income and credit exposure influenced customer risk levels.
- Education level contributed meaningfully to creditworthiness assessment.

### Risk Segmentation Findings

- Low Risk Segment Default Rate: 1.89%
- Medium Risk Segment Default Rate: 3.97%
- High Risk Segment Default Rate: 7.60%
- Very High Risk Segment Default Rate: 18.83%

### Model Performance

- XGBoost delivered the strongest overall performance with a ROC-AUC of approximately 0.755.
- Logistic Regression remained highly competitive and offered superior interpretability.
- Random Forest provided limited improvement relative to Logistic Regression.

### Business Recommendations

- Automatically approve Low Risk applicants.
- Apply tighter credit limits to Medium Risk applicants.
- Route High Risk applicants for manual underwriting review.
- Reject or heavily restrict Very High Risk applicants.
